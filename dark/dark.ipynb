```python
from ultralytics import YOLO
import cv2
```

```python
from pathlib import Path
import shutil
```

```python
BASE_DIR = Path(r'B:\datasets\dark')
```

```python
model_dir = Path.cwd() / 'yolo training'
model = YOLO(model_dir / "yolo11l.pt")
```

```python
def detect(frame):
    results = model(frame)
    frame_after = frame.copy()
    for i,result in enumerate (results):
        print(i,result)
        xyxy = result.boxes.xyxy
        names = [result.names[cls.item()] for cls in result.boxes.cls.int()]
        for name, [x1,y1,x2,y2] in zip(names,xyxy):
            cv2.rectangle(frame_after, (int(x1),int(y1)), (int(x2),int(y2)), (255, 0, 0), 3)
    return frame_after
```

```python
img = cv2.imread(str(BASE_DIR / 'image' / '100.png'))
img2 = detect(img)
cv2.imshow('frame', img2)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import numpy as np

dft = cv2.dft(np.float32(cv2.cvtColor(img,cv2.COLOR_BGRA2GRAY)),flags = cv2.DFT_COMPLEX_OUTPUT)
dft_shift = np.fft.fftshift(dft)
magnitude_spectrum = 20*np.log(cv2.magnitude(dft_shift[:,:,0],dft_shift[:,:,1]))

cv2.imshow('frame', magnitude_spectrum)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
